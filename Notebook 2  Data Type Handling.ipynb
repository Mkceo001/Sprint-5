#  Data Type Handling
Data Type Handling is the process of identifying, checking, and converting data into the correct format before performing data analysis or building Machine Learning models.

In simple words:

Data Type Handling means ensuring that every column contains the correct type of information.

## Identifying Data Types
Pandas provides several methods to identify data types.

### Important methods:
Method	Purpose
- df.dtypes	Shows column data types
- df.info()	Shows complete dataset information
- df.select_dtypes()	Selects columns by data type
- type()	Identifies Python object type

### Example code:

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Arun', 'Kumar', 'Priya'],
    'Age': [25, 30, 28],
    'Salary': [25000.5, 30000.0, 28000.5],
    'Active': [True, False, True]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nColumn Data Types:")
print(df.dtypes)

print("\nComplete Dataset Information:")
df.info()
```

## Types of Data
There are five important data types to understand.
### Numerical Data
Numerical Data contains numbers that can be used for mathematical calculations.

There are two main categories:

#### Integer (int)

Integer values are whole numbers without decimal points.

##### Examples:
- Age = 25
- Experience = 3
- Number of employees = 100

```python
age = 25

print(age)
print(type(age))
```

#### Float (float)

Float values contain decimal points.

##### Examples:
- Salary = 25000.50
- Height = 5.8
- Temperature = 36.5

```python
salary = 25000.50

print(salary)
print(type(salary))
```

### Categorical Data
Categorical Data represents groups, labels, or categories rather than numerical measurements.

##### Examples:
- Gender
- City
- Department
- Product Category
- Payment Method

```python
data = {
    'Name': ['Arun', 'Kumar', 'Priya'],
    'City': ['Chennai', 'Madurai', 'Salem'],
    'Department': ['IT', 'HR', 'Finance']
}

df = pd.DataFrame(data)

print(df)
print("\nData Types:")
print(df.dtypes)
```

### Boolean Data
Boolean Data contains only two logical values:
- True
- False
#### Examples:
- Is employee active?
- Is payment completed?
- Is customer eligible?

```python
data = {
    'Employee': ['Arun', 'Kumar', 'Priya'],
    'Is_Active': [True, False, True]
}

df = pd.DataFrame(data)

print(df)
print(df.dtypes)
```

### Date/Time Data
Date/Time Data represents dates and times.

##### Examples:
- Joining date
- Order date
- Birth date
- Payment date
- Login time

```python
dates = pd.Series([
    '2024-01-10',
    '2024-02-15',
    '2024-03-20'
])

print(dates)
print(dates.dtype)
```

### String Data
String Data contains text values.

##### Examples:
- Customer names
- Addresses
- Phone numbers
- Email addresses
- Product descriptions

```python
name = "Mugeshkannan"

print(name)
print(type(name))
```

## Converting Data Types
Data Type Conversion means changing a column from one data type into another.

#### Example:
- String → Integer
- String → Float
- String → DateTime
- Integer → Float
- Numeric → Categorical

## Using astype()
astype() is used to convert a Pandas column into a specific data type.

### Example :
Integer → Float

```python
data = {
    'Age': [25, 30, 35, 40]
}

df = pd.DataFrame(data)

print("Before Conversion:")
print(df.dtypes)

df['Age'] = df['Age'].astype(float)

print("\nAfter Conversion:")
print(df.dtypes)

print(df)
```

##  Using pd.to_numeric()
pd.to_numeric() converts values into numerical data types.

It is particularly useful when numbers are stored as strings.

### Example: 
String → Numeric

```python
data = {
    'Salary': ['20000', '30000', '40000', '50000']
}

df = pd.DataFrame(data)

print("Before Conversion:")
print(df.dtypes)

df['Salary'] = pd.to_numeric(df['Salary'])

print("\nAfter Conversion:")
print(df.dtypes)

print(df)
```

## Using pd.to_datetime()
pd.to_datetime() converts string values into proper date and time values.

### Example: 
String → Date

```python
data = {
    'Joining_Date': [
        '2024-01-10',
        '2024-02-15',
        '2024-03-20'
    ]
}

df = pd.DataFrame(data)

print("Before Conversion:")
print(df.dtypes)

df['Joining_Date'] = pd.to_datetime(
    df['Joining_Date']
)

print("\nAfter Conversion:")
print(df.dtypes)

print(df)
```

## Handling Invalid Conversions
Sometimes a dataset contains values that cannot be converted into the required data type.

### Example:

```python
data = {
    'Salary': ['20000', '30000', 'Invalid', '50000']
}

df = pd.DataFrame(data)

df['Salary'] = pd.to_numeric(
    df['Salary'],
    errors='coerce'
)

print(df)
```

## Handling Invalid Date Conversions
Date columns can also contain incorrect formats.

### Example:

```python
data = {
    'Date': [
        '2024-01-10',
        '2024-02-15',
        'Invalid Date',
        '2024-04-20'
    ]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(
    df['Date'],
    errors='coerce'
)

print(df)

print("\nInvalid Dates:")
print(df[df['Date'].isna()])
```

# Feature Selection

## What is Feature Selection?
Feature Selection is the process of selecting the most important and useful input columns (features) from a dataset for building a Machine Learning model.

In simple words, Feature Selection means choosing only the necessary columns and removing unnecessary columns from a dataset.

## Why Feature Selection?
Feature Selection helps Machine Learning models work more efficiently.

Main benefits:
1. Improves model performance.
2. Reduces training time.
3. Removes unnecessary information.
4. Reduces overfitting.
5. Makes the model easier to understand.
6. Reduces computational complexity.

## ypes of Features
There are three important types of features.
### A. Relevant Features
Relevant features are input columns that provide useful information for predicting the target variable.

#### Example:

Predicting house prices:
- House area
- Location
- Number of bedrooms
These features influence house prices.
### B. Irrelevant Features
Irrelevant features are columns that do not provide useful information for predicting the target variable.

#### Example:
- Customer ID
- Random reference number
- Unrelated serial number
These features usually do not improve prediction.
### C. Redundant Features
Redundant features are columns that contain almost the same information as other features.


## Correlation-Based Selection
Correlation measures the relationship between two numerical variables.

Its value ranges from -1 to +1.
### Example
- Study hours and marks: Positive correlation.

- Product price and demand: Often negative correlation.

```python
import pandas as pd

data = {
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [20, 35, 50, 70, 90],
    'Student_ID': [101, 102, 103, 104, 105]
}

df = pd.DataFrame(data)

correlation = df.corr()

print(correlation)
```

## Variance Threshold
Variance Threshold is a feature selection technique that removes features with very little variation.

In simple words, if a column contains almost the same value in every row, it may not provide useful information.
### Example:

```python
from sklearn.feature_selection import VarianceThreshold
import pandas as pd

data = pd.DataFrame({
    'Age': [20, 25, 30, 35],
    'Salary': [20000, 30000, 40000, 50000],
    'Constant': [1, 1, 1, 1]
})

selector = VarianceThreshold(threshold=0)

result = selector.fit_transform(data)

print(result)
print(selector.get_feature_names_out())
```

## Univariate Feature Selection
Univariate Feature Selection evaluates each feature individually to determine how useful it is for predicting the target variable.

### Example:

Predicting student examination results:
- Study Hours
- Attendance
- Previous Marks
Each feature is evaluated separately.

```python
from sklearn.datasets import load_iris
from sklearn.feature_selection import SelectKBest, f_classif

iris = load_iris()

X = iris.data
y = iris.target

selector = SelectKBest(score_func=f_classif, k=2)

X_selected = selector.fit_transform(X, y)

print("Original Features:", X.shape)
print("Selected Features:", X_selected.shape)
print("Selected Columns:",
      iris.feature_names)
```

## Mutual Information
Mutual Information measures how much information one feature provides about the target variable.

In simple words, it helps identify how useful a feature is for predicting the target, including certain non-linear relationships.

### Example:

Predicting whether a customer purchases a product:
- Age
- Income
- Website Visits
- Random ID

```python
from sklearn.datasets import load_iris
from sklearn.feature_selection import mutual_info_classif

iris = load_iris()

X = iris.data
y = iris.target

scores = mutual_info_classif(
    X, y, random_state=42
)

for name, score in zip(iris.feature_names, scores):
    print(name, ":", score)
    
```

## Recursive Feature Elimination (RFE)
RFE stands for Recursive Feature Elimination.

It is a technique that repeatedly trains a model and removes the least important features until the required number of features remains.

Working process:
1. Train a Machine Learning model.
2. Calculate feature importance.
3. Remove the least important feature.
4. Train the model again.
5. Repeat until the required features remain.

```python
from sklearn.datasets import load_iris
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()

X = iris.data
y = iris.target

model = RandomForestClassifier(
    random_state=42
)

selector = RFE(
    estimator=model,
    n_features_to_select=2
)

selector.fit(X, y)

print("Selected Features:")

for name, selected in zip(
    iris.feature_names,
    selector.support_
):
    if selected:
        print(name)
```

## Feature Importance
Feature Importance identifies which input features contribute most to a model's predictions.

### Example:

Predicting house prices:

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()

X = iris.data
y = iris.target

model = RandomForestClassifier(
    random_state=42
)

model.fit(X, y)

importance = model.feature_importances_

for name, score in zip(
    iris.feature_names,
    importance
):
    print(name, ":", score)
```

```python
import pandas as pd

from sklearn.feature_selection import (
    VarianceThreshold,
    SelectKBest,
    f_regression
)

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Create dataset
data = {
    'Study_Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Attendance': [40, 50, 60, 65, 75, 80, 90, 95],
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Constant': [1, 1, 1, 1, 1, 1, 1, 1],
    'Marks': [25, 35, 45, 50, 65, 70, 85, 95]
}

df = pd.DataFrame(data)

# Separate input and target
X = df.drop('Marks', axis=1)
y = df['Marks']

# Split data first
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Remove constant feature
variance_selector = VarianceThreshold(threshold=0)

X_train_var = variance_selector.fit_transform(X_train)
X_test_var = variance_selector.transform(X_test)

selected_columns = variance_selector.get_feature_names_out()

X_train_var = pd.DataFrame(
    X_train_var, columns=selected_columns,
    index=X_train.index
)

X_test_var = pd.DataFrame(
    X_test_var, columns=selected_columns,
    index=X_test.index
)

# Select top 2 features
selector = SelectKBest(
    score_func=f_regression,
    k=2
)

X_train_selected = selector.fit_transform(
    X_train_var, y_train
)

X_test_selected = selector.transform(X_test_var)

print("Selected Features:",
      selected_columns[selector.get_support()])

# Train model
model = LinearRegression()

model.fit(X_train_selected, y_train)

# Predict
predictions = model.predict(X_test_selected)

print("Actual Marks:", list(y_test))
print("Predicted Marks:", predictions.round(2))

print("R2 Score:", r2_score(y_test, predictions))
```

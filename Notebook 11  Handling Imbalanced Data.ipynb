# Handling Imbalanced Data

## What is Class Imbalance?
Class Imbalance occurs when one class has significantly more records than another class in a dataset.

In simple words, one category has many data points, while another category has very few data points.
#### Real-world example:
Bank fraud detection

## Why is Class Imbalance a Problem?
Class imbalance can cause a Machine Learning model to favor the majority class.

#### For example:

Suppose we train a fraud detection model with:
- 950 normal transactions.
- 50 fraud transactions.
The model may learn to predict every transaction as normal because normal transactions are more common.

As a result:
- It correctly identifies most normal transactions.
- It fails to identify fraudulent transactions.
- Its accuracy looks high, but its real-world usefulness is poor.
This is why we need special techniques to handle imbalanced datasets.
## 4. Why Accuracy Alone Can Be Misleading
This is the most important concept in this notebook.

- 950 normal transactions.

## echniques for Handling Imbalanced Data
There are two main approaches:
#### 1. Undersampling

Reduce the number of records in the majority class.

##### Example: 
Reduce 950 normal records to 50.

#### 2. Oversampling

Increase the number of records in the minority class.

##### Example: 
Increase 50 fraud records to 950.

## A. Random Oversampling
Random Oversampling increases the minority class by randomly duplicating existing records.


## B. Random Undersampling
Random Undersampling reduces the majority class by randomly removing records.
## C. MOTE (Synthetic Minority Over-sampling Technique)
SMOTE creates new synthetic minority-class examples instead of simply duplicating existing records.

In simple words, it creates artificial data points based on similarities between existing minority-class records.

#### Example:

Suppose we have two fraud records:
- Record A: Amount = ₹1,000
- Record B: Amount = ₹2,000
## D. Borderline-SMOTE
Borderline-SMOTE is an improved version of SMOTE.

It focuses on minority-class examples near the decision boundary, where the two classes are difficult to separate.

In simple words, it creates more synthetic examples around the confusing areas between the majority and minority classes.
## E. Class Weights
Class Weights tell the Machine Learning model to give more importance to the minority class during training.

Instead of changing the dataset, we change how much the model penalizes mistakes on different classes.

#### Example:
Class	Records	Importance
Normal	950	Lower
Fraud	50	Higher

## When to Use Each Technique?
Technique	When to use

Random Oversampling:	Small dataset; preserve all majority records

Random Undersampling:	Very large majority class; reduce training time

SMOTE:	Numerical data with enough minority examples

Borderline-SMOTE:	Minority examples near difficult decision boundaries

Class Weights:	When you don't want to modify dataset size


#### Practical recommendation:
- Small dataset: Try Random Oversampling or SMOTE.
- Large dataset: Try Random Undersampling.
- Difficult classification boundary: Try Borderline-SMOTE.
- Want to preserve original dataset: Try Class Weights.

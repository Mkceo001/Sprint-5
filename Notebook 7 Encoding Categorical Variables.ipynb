# Encoding Categorical Variables

## What is Categorical Encoding?
Categorical encoding is the process of converting text-based categorical values into numerical values so that Machine Learning algorithms can understand and process them.

```python
import pandas as pd

df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female']
})

print(df)
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['Gender_Encoded'] = encoder.fit_transform(df['Gender'])

print(df)
```

## Types of categorical variables
Before selecting an encoding technique, understand the type of categorical variable.
#### A. Nominal Variables
Nominal variables contain categories without any natural order.

##### Examples:
- Gender: Male, Female
- City: Chennai, Madurai, Trichy
- Department: HR, IT, Finance
- Colour: Red, Blue, Green
When to use: When categories have no ranking or meaningful order. One-Hot Encoding is commonly suitable.
#### B. Ordinal Variables
Ordinal variables contain categories with a meaningful order or ranking.

##### Examples:
- Education: School, Undergraduate, Postgraduate
- Satisfaction: Low, Medium, High
- Size: Small, Medium, Large
When to use: When the order of categories is important. Ordinal Encoding is usually suitable.
#### C. Binary Variables
Binary variables have only two possible categories.

##### Examples:
- Yes / No
- True / False
- Active / Inactive
- Purchased / Not Purchased
When to use: When a feature contains exactly two meaningful categories. You can map them to 0 and 1.

## Label Encoding
Label Encoding converts each category into a numerical value.
### Example:

```python
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    'Department': ['HR', 'IT', 'Finance', 'HR', 'IT']
})

encoder = LabelEncoder()

df['Department_Encoded'] = encoder.fit_transform(
    df['Department']
)

print(df)
```

## Ordinal Encoding
Ordinal Encoding converts categories into numerical values while preserving their meaningful order.

### Example:

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'Satisfaction': ['Low', 'High', 'Medium', 'Low', 'High']
})

encoder = OrdinalEncoder(
    categories=[['Low', 'Medium', 'High']]
)

df['Satisfaction_Encoded'] = encoder.fit_transform(
    df[['Satisfaction']]
)

print(df)
```

## One-Hot Encoding
One-Hot Encoding creates a separate binary column for each category.

### Example:

Original data:
- City
- Chennai
- Madurai
- Trichy

```python
import pandas as pd

df = pd.DataFrame({
    'City': ['Chennai', 'Madurai', 'Trichy', 'Chennai']
})

encoded_df = pd.get_dummies(
    df,
    columns=['City'],
    dtype=int
)

print(encoded_df)
```

## Dummy Variables
Dummy variables are binary columns representing categories.

They are closely related to One-Hot Encoding. The common difference is that dummy encoding may drop one category to use it as the reference category.

### Example:

Original:
- Gender
- Male
- Female
- Male

```python
import pandas as pd

df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male', 'Female']
})

dummy_df = pd.get_dummies(
    df,
    columns=['Gender'],
    drop_first=True,
    dtype=int
)

print(dummy_df)
```

## Frequency Encoding
Frequency Encoding replaces each category with how frequently it appears in the dataset.

### Example:

Original data:
- City
- Chennai
- Chennai
- Madurai
- Trichy
- Chennai
- Madurai

```python
import pandas as pd

df = pd.DataFrame({
    'City': [
        'Chennai', 'Chennai', 'Madurai',
        'Trichy', 'Chennai', 'Madurai'
    ]
})

frequency = df['City'].value_counts()

df['City_Frequency'] = df['City'].map(frequency)

print(df)
```

## Target Encoding
Target Encoding replaces each category with a statistic calculated from the target variable, commonly its mean for a regression or binary classification target.

### Example: 
Predict whether a customer will purchase a product.

```python
import pandas as pd

df = pd.DataFrame({
    'City': [
        'Chennai', 'Chennai', 'Chennai',
        'Madurai', 'Madurai', 'Trichy'
    ],
    'Purchased': [1, 1, 0, 0, 0, 1]
})

target_means = df.groupby('City')['Purchased'].mean()

df['City_Target_Encoded'] = df['City'].map(target_means)

print(df)
```

## Handling Unknown Categories
An unknown category is a category that appears in new data but was not present when the encoder was trained.

### Example:

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

train = pd.DataFrame({
    'City': ['Chennai', 'Madurai', 'Trichy']
})

test = pd.DataFrame({
    'City': ['Chennai', 'Coimbatore']
})

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

train_encoded = encoder.fit_transform(train[['City']])
test_encoded = encoder.transform(test[['City']])

print("Training data:")
print(train_encoded)

print("\nNew data:")
print(test_encoded)
```

## High-Cardinality Categories
High-cardinality features contain a large number of unique categories.

### Example:

```python
import pandas as pd

df = pd.DataFrame({
    'City': [
        'Chennai', 'Chennai', 'Chennai',
        'Madurai', 'Trichy', 'Salem',
        'Erode', 'Chennai'
    ]
})

counts = df['City'].value_counts()

df['City_Grouped'] = df['City'].where(
    df['City'].map(counts) >= 2,
    'Other'
)

print(df)
```

# Data Transformation

## What is Data Transformation?
Data Transformation is a data preprocessing technique used to change the format, structure, or distribution of data to make it more suitable for machine learning algorithms.

In simple words, Data Transformation means converting data from one form into another form to improve its usability and analysis.

```python
Salary = [15000, 18000, 20000, 25000, 30000, 500000]
```

## Why Transform Data?
Data transformation is useful for the following reasons:
1. Reduces skewness in numerical data.
2. Makes data distribution more balanced.
3. Helps certain machine learning algorithms perform better.
4. Reduces the influence of extreme values on the distribution.
5. Helps identify patterns and relationships.
6. Improves the effectiveness of statistical analysis.

## Understanding Skewness
Before learning transformation techniques, we need to understand skewness.

Skewness measures how much a data distribution is asymmetrical.

There are three types:
#### A. Normal Distribution
- Data is approximately symmetrical.
- Mean and median are approximately equal.
- Skewness is close to zero.
#### B. Positive Skewness (Right-Skewed)
- The tail extends towards the right.
- Most values are smaller.
- A few values are extremely large.
- Mean is usually greater than median.
##### Example:
 Salaries, house prices and income.
 #### C. Negative Skewness (Left-Skewed)
- The tail extends towards the left.
- Most values are larger.
- A few values are extremely small.
- Mean is usually less than median.
##### Example:
An examination where most students score highly but a few score very low.


## Log Transformation
Log Transformation is used to reduce positive skewness by applying the logarithm to numerical values.

It is especially useful when a dataset contains very large values.
### Formula


We commonly use np.log1p() in Python. Adding 1 allows zero values.
### Example:

```python
import numpy as np
import pandas as pd

# Original data
data = pd.Series([10, 20, 30, 40, 50, 1000])

# Log transformation
log_data = np.log1p(data)

print("Original Data:")
print(data.values)

print("\nLog Transformed Data:")
print(log_data.round(2).values)
```

## Square Root Transformation
Square Root Transformation applies the square root to each numerical value.

It is useful for reducing moderate positive skewness.
### Formula

### Example:

```python
import numpy as np

data = np.array([4, 9, 16, 25, 36, 100])

sqrt_data = np.sqrt(data)

print("Original Data:")
print(data)

print("\nSquare Root Data:")
print(sqrt_data)
```

## Power Transformation
Power Transformation changes the distribution of data using mathematical power functions.

It helps reduce skewness and make data more suitable for statistical and machine learning algorithms.

Two commonly used power transformations are:
1. Box-Cox Transformation
2. Yeo-Johnson Transformation
We can implement both using Scikit-learn.

```python
from sklearn.preprocessing import PowerTransformer
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
    [1000]
])

transformer = PowerTransformer(
    method='yeo-johnson'
)

transformed_data = transformer.fit_transform(data)

print(transformed_data.round(3))
```

## Box-Cox Transformation
Box-Cox Transformation is a statistical technique used to make positively skewed data more normally distributed.

It uses a parameter called lambda (λ) to determine the best transformation.
### Formula

### Example:

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50],
    [1000]
])

transformer = PowerTransformer(
    method='box-cox'
)

boxcox_data = transformer.fit_transform(data)

print("Box-Cox Transformed Data:")
print(boxcox_data.round(3))

print("Lambda Value:")
print(transformer.lambdas_)
```

## Yeo-Johnson Transformation
Yeo-Johnson Transformation is similar to Box-Cox, but it can handle:
- Positive values
- Negative values
- Zero values
This makes it more flexible.
### Example:

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

data = np.array([
    [-20],
    [-10],
    [0],
    [10],
    [20],
    [100]
])

transformer = PowerTransformer(
    method='yeo-johnson'
)

yeo_data = transformer.fit_transform(data)

print("Yeo-Johnson Transformed Data:")
print(yeo_data.round(3))

print("Lambda Value:")
print(transformer.lambdas_)
```

## Handling Skewed Data
Handling skewed data means reducing the imbalance in a numerical distribution.

#### Example:

Imagine an online shopping company analyzing customer purchases.

Most customers purchase products worth ₹500–₹2,000, while a few customers purchase products worth ₹50,000.

This creates positive skewness.

```python
import pandas as pd

data = pd.Series([
    10, 20, 30, 40, 50,
    60, 70, 80, 100, 1000
])

print("Skewness:", data.skew())
```

## Normalization
Normalization converts numerical values into a common range, usually between 0 and 1.
#### Formula

### Example:

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(data)

print(normalized_data)
```

## Standardization
Standardization converts numerical data into a distribution with:
- Mean approximately equal to 0.
- Standard deviation approximately equal to 1.
#### Formula

#### Example

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

scaler = StandardScaler()

standardized_data = scaler.fit_transform(data)

print(standardized_data)
```

# Feature Scaling

## What is Feature Scaling?
Feature Scaling is a data preprocessing technique used to bring numerical features into a similar range or scale.

In machine learning, different features may have different measurement units and ranges. If we use these values directly, some algorithms may give more importance to features with larger numerical values.

### Example:

Imagine we are building a machine learning model to predict house prices.

## Why Feature Scaling?
Feature scaling is important because:
1. Prevents large numerical values from dominating smaller values.
2. Improves machine learning model performance.
3. Helps algorithms converge faster.
4. Makes distance calculations more meaningful.
5. Improves the performance of certain machine learning algorithms.


## Scale-Sensitive Algorithms
Scale-sensitive algorithms are machine learning algorithms that are affected by the numerical range of input features.

These algorithms usually require feature scaling.

### Examples:
Algorithm	               Why scaling is useful

KNN      :                 	Uses distance calculations

K-Means :                 	Uses distance between points

SVM	:                       Uses distances and margins

Logistic :                  Regression	Helps optimization

Linear Regression with regularization	: Helps balance feature penalties

Neural Networks	:               Helps stable and faster training

PCA	:                    Depends on feature variance

## Scale-Insensitive Algorithms
These algorithms generally do not require feature scaling because their decisions are not based on the magnitude of numerical features in the same way.

### Examples:
- Decision Tree
- Random Forest
- XGBoost
- Gradient Boosting
- Extra Trees

## Normalization
Normalization transforms numerical values into a specified range, commonly between 0 and 1.
### Formula


### Example

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

normalized = (data - data.min()) / (data.max() - data.min())

print(normalized)
```

## Standardization
Standardization transforms data so that it has:
- Mean approximately equal to 0.
- Standard deviation approximately equal to 1.
### Formula


Where:
- X = Original value
- μ = Mean
- σ = Standard deviation
### Example:

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

data = np.array([[10], [20], [30], [40], [50]])

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
```

## Min-Max Scaling
Min-Max Scaling converts values into a specified range, usually 0 to 1.

It is implemented using MinMaxScaler.
### Example:

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
```

## StandardScaler
StandardScaler transforms features using their mean and standard deviation.
### Example:

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

scaler = StandardScaler()

result = scaler.fit_transform(data)

print(result)
```

## RobustScaler
RobustScaler is useful when the dataset contains outliers.

Instead of using mean and standard deviation, it uses:
- Median
- Interquartile Range (IQR)
### Formula


Where:
- Q1 = First quartile (25%)
- Q3 = Third quartile (75%)
- IQR = Q3 − Q1
### Example:

```python
from sklearn.preprocessing import RobustScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [1000]
])

scaler = RobustScaler()

result = scaler.fit_transform(data)

print(result)
```

## MaxAbsScaler
MaxAbsScaler scales values by dividing each feature by its maximum absolute value.
### Formula



The resulting values generally fall between -1 and 1.
### Example

```python
from sklearn.preprocessing import MaxAbsScaler
import numpy as np

data = np.array([
    [10],
    [20],
    [30],
    [40],
    [50]
])

scaler = MaxAbsScaler()

result = scaler.fit_transform(data)

print(result)
```

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    RobustScaler,
    MaxAbsScaler
)

# Original dataset
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 25000, 30000, 35000, 100000]
})

print("Original Data:")
print(df)

# Apply different scaling techniques
scalers = {
    "StandardScaler": StandardScaler(),
    "MinMaxScaler": MinMaxScaler(),
    "RobustScaler": RobustScaler(),
    "MaxAbsScaler": MaxAbsScaler()
}

for name, scaler in scalers.items():

    scaled_data = scaler.fit_transform(df)

    scaled_df = pd.DataFrame(
        scaled_data,
        columns=df.columns
    )

    print("\n", name)
    print(scaled_df.round(3))
```

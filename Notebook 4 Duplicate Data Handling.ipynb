#  Duplicate Data Handling
Duplicate records are rows in a dataset that contain the same information or represent the same real-world record more than once.

### Example
Imagine an online shopping company has stored the same customer order twice. This can lead to incorrect sales reports and customer counts.

```python
import pandas as pd

data = {
    'Customer_ID': [101, 102, 103, 101, 104],
    'Name': ['Arun', 'Priya', 'Kumar', 'Arun', 'Divya'],
    'City': ['Chennai', 'Madurai', 'Salem', 'Chennai', 'Trichy'],
    'Amount': [500, 700, 300, 500, 900]
}

df = pd.DataFrame(data)

print(df)
```

## Types of Duplicate Records
### Exact Duplicates
An exact duplicate occurs when every column in two or more rows contains the same values.

#### Example:
Customer_ID	Name	Amount

101	Arun	500

101	Arun	500
### Partial Duplicates
A partial duplicate occurs when some selected columns contain the same values, but other columns are different.

#### Example:
Customer_ID	Name	City

101	Arun	Chennai

101	Arun	Coimbatore

## Identifying Duplicates Using duplicated()
The duplicated() function identifies duplicate rows.
### Example:

```python
print(df.duplicated())
```

```python
duplicates = df[df.duplicated()]

print(duplicates)
```

## Removing Duplicates Using drop_duplicates()
The drop_duplicates() function removes duplicate rows.
### Example:

```python
clean_df = df.drop_duplicates()

print(clean_df)
```

### Duplicate Detection Based on Selected Columns
Sometimes, we need to check duplicates using only specific columns rather than every column.

#### Example:

```python
data = {
    'Customer_ID': [101, 102, 101, 103],
    'Name': ['Arun', 'Priya', 'Arun', 'Kumar'],
    'City': ['Chennai', 'Madurai', 'Coimbatore', 'Salem']
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.duplicated(subset=['Customer_ID']))
```

```python
duplicate_customers = df[
    df.duplicated(subset=['Customer_ID'], keep=False)
]

print(duplicate_customers)
```

```python
clean_df = df.drop_duplicates(
    subset=['Customer_ID'],
    keep='first'
)

print(clean_df)
```

## Handling Duplicate Records
Different situations require different handling methods.
#### Method 1: Remove exact duplicates
Use when records are genuinely repeated and all relevant columns match.

```python
df = df.drop_duplicates()
```

#### Method 2: Keep the latest record
Suppose customer information is updated over time. We may want to keep the latest record instead of the first.

```python
data = {
    'Customer_ID': [101, 102, 101, 103],
    'Name': ['Arun', 'Priya', 'Arun', 'Kumar'],
    'City': ['Chennai', 'Madurai', 'Coimbatore', 'Salem'],
    'Updated_Date': [
        '2024-01-10',
        '2024-01-12',
        '2024-02-15',
        '2024-01-20'
    ]
}

df = pd.DataFrame(data)

df['Updated_Date'] = pd.to_datetime(df['Updated_Date'])

clean_df = (
    df.sort_values('Updated_Date')
      .drop_duplicates(subset=['Customer_ID'], keep='last')
)

print(clean_df)
```

#### Method 3: Review duplicates instead of deleting them
Sometimes, we should inspect possible duplicates manually.

```python
possible_duplicates = df[
    df.duplicated(subset=['Customer_ID'], keep=False)
].sort_values('Customer_ID')

print(possible_duplicates)
```

## usiness Rules for Duplicate Removal
Business rules define when two records should be treated as duplicates and which record should be retained.

##### Business situation	Possible rule

Customer records	Same Customer_ID

Employee records	Same Employee_ID

Product records	Same Product_ID

Bank transactions	Same Transaction_ID

Online orders	Same Order_ID

Updated customer details	Keep latest Updated_Date
